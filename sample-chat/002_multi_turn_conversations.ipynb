```python
# Load env variables
from dotenv import load_dotenv

load_dotenv()
```

```python
# Create an API Client

from anthropic import Anthropic

client = Anthropic()
model = "claude-sonnet-4-5"
max_tokens = 1000

def add_user_message(messages, content):
    messages.append({"role": "user", "content": content})

def add_assistant_message(messages, content):
    messages.append({"role": "assistant", "content": content})

def chat(messages):
    response = client.messages.create(
        model=model,
        messages=messages,
        max_tokens=max_tokens,
    )
    return response.content[0].text
```

```python
# Start with empty message list

messages = []

# Add initial user question
add_user_message(messages, "What is quantum computing? Answer in one sentence.")

# Get the answer from the model
answer = chat(messages)

#Add Claude's answer to the messages list
add_assistant_message(messages, answer)

# Add a follow-up request

add_user_message(messages, "Write me another sentence that expands on your previous answer.")

final_answer = chat(messages)

add_assistant_message(messages, final_answer)

messages
```
